```python
# create function to extract text from all pdf files  
def extract_pdf_text(filepath):

    # empty list to store extracted text  
    page_texts = []

    # open folder
    with pdfplumber.open(filepath) as pdf:

        # go through every page in the PDF one at a time
        for page in pdf.pages:
            # extract the text from the current page
            text = page.extract_text()
            
            # if there is text add it to the empty list we made 
            if text:
                page_texts.append(text)

            # Extract tables from the page
            tables = page.extract_tables()

            # Go through each table
            for table_number, table in enumerate(tables, start=1):

                page_texts.append(
                    f"\n--- Table {table_number} ---\n"
                )

                # Go through each row
                for row in table:

                    # Remove None values and convert cells to strings
                    clean_row = [
                        str(cell).strip() if cell else ""
                        for cell in row
                    ]

                    # Join each cell with a separator
                    row_text = " | ".join(clean_row)

                    # Add the row to the document
                    page_texts.append(row_text)
    
    # new line and join all the text in the list together 
    return "\n".join(page_texts)
```

```python
def extract_docx_text(filepath):

    paragraphs = []

    document = Document(filepath)

    for paragraph in document.paragraphs:

        text = paragraph.text

        if text:
            paragraphs.append(text)

    return "\n".join(paragraphs)
```

```python
# give the path of the assessment briefs folder 
folder_path = Path("assessment_briefs")

# find every file ending in .pdf inside assessment_briefs and add to a list.
pdf_files = list(folder_path.glob("*.pdf"))

#find every file ending in .docx inside assessment_briefs and add to a list.
docx_files = list(folder_path.glob("*.docx"))
```

```python
# go through each PDF file in my pdf_files list, one at a time
for pdf_file in pdf_files:

    # display which PDF you're currently processing
    print(f"\n--- {pdf_file.name} ---")#

    # run the extract_pdf_text function and take the current PDF and extract all of its text
    text = extract_pdf_text(pdf_file)

    print(text)
```
